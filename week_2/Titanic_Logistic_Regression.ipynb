```python
# Titanic Survival Prediction using Logistic Regression

## Objective
##Predict whether a passenger survived the Titanic disaster using Logistic Regression.
import pandas as pd
## Data Preparation
#The cleaned Titanic dataset from Week 1 is used. PassengerId, Name, and Ticket are removed because they are identifiers/text fields rather than features used in this basic model.
new_dataset=pd.read_csv('titanic_cleaned.csv')
new_dataset.head(5)
```

```python
new_dataset.info()
```

```python
x = new_dataset.drop(['PassengerId', 'Name', 'Ticket','Survived'], axis=1)
y = new_dataset['Survived']
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=42)
from sklearn.linear_model import LogisticRegression
model_training=LogisticRegression(max_iter=1000)
## Model Training
##The dataset is split into 80% training and 20% testing data. A Logistic Regression model is trained on the training data.
model_training.fit(x_train, y_train)
predictions=model_training.predict(x_test)
## Evaluation
##The model is evaluated using accuracy score and a confusion matrix.
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test,predictions)
accuracy
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, predictions)
print(cm)
```
