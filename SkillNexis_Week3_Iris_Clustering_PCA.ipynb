```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
iris_dataset= pd.read_csv('Iris.csv')
X = iris_dataset[['SepalLengthCm',
                  'SepalWidthCm',
                  'PetalLengthCm',
                  'PetalWidthCm']]
y = iris_dataset['Species']
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)
clusters = kmeans.labels_
print(kmeans.cluster_centers_)
```

```python
plt.scatter(X['SepalLengthCm'], X['SepalWidthCm'], c=clusters)
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('K-Means Clusters')
plt.show()
```

```python
comparison = pd.crosstab(y, clusters)
print(comparison)
```

```python
#PCA 
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
print(X_pca.shape)
print(pca.explained_variance_ratio_)
print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total variance explained:", pca.explained_variance_ratio_.sum())
#PCA reduced the Iris dataset from 4 dimensions to 2 principal components. The first principal component explains 72.77% of the variance, while the second explains 23.03%. Together, the two components preserve approximately 95.80% of the total variance.
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=clusters)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('K-Means Clusters after PCA')

plt.show()
```

```python
Conclusion: K-Means successfully grouped the Iris dataset into three clusters based on flower measurements. The clustering showed strong separation of Iris-setosa, while some overlap occurred between Iris-versicolor and Iris-virginica. PCA reduced the four original features to two #principal components while retaining approximately 95.80% of the total variance, making the clusters easier to visualize in two dimensions.
```
